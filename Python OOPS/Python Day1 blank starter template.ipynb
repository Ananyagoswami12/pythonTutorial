```python
from abc import ABC, abstractmethod
from datetime import datetime


# -----------------------------
# Abstract Base Class
# -----------------------------
class Account(ABC):

    bank_name = "Global Bank"

    def __init__(self, account_number, holder_name, balance=0):
        self._account_number = account_number     # Encapsulation
        self._holder_name = holder_name
        self.__balance = balance                  # Private variable

    # Getter Method
    def get_balance(self):
        return self.__balance

    # Protected method
    def _update_balance(self, amount):
        self.__balance += amount

    @abstractmethod
    def calculate_interest(self):
        pass

    def deposit(self, amount):
        if amount > 0:
            self._update_balance(amount)
            print(f"{amount} deposited successfully.")

    def withdraw(self, amount):
        if amount <= self.__balance:
            self._update_balance(-amount)
            print(f"{amount} withdrawn successfully.")
        else:
            print("Insufficient balance.")

    @classmethod
    def get_bank_name(cls):
        return cls.bank_name

    @staticmethod
    def validate_amount(amount):
        return amount > 0


# -----------------------------
# Savings Account
# -----------------------------
class SavingsAccount(Account):

    interest_rate = 0.04

    def calculate_interest(self):
        return self.get_balance() * self.interest_rate


# -----------------------------
# Current Account
# -----------------------------
class CurrentAccount(Account):

    overdraft_limit = 5000

    # Polymorphism (Method Overriding)
    def withdraw(self, amount):
        if amount <= self.get_balance() + self.overdraft_limit:
            self._update_balance(-amount)
            print(f"{amount} withdrawn from Current Account.")
        else:
            print("Overdraft limit exceeded.")

    def calculate_interest(self):
        return 0


# -----------------------------
# Composition Example
# -----------------------------
class Transaction:

    def __init__(self, transaction_type, amount):
        self.transaction_type = transaction_type
        self.amount = amount
        self.timestamp = datetime.now()

    def show_transaction(self):
        print(
            f"{self.transaction_type} | "
            f"{self.amount} | "
            f"{self.timestamp}"
        )


# -----------------------------
# Customer Class
# -----------------------------
class Customer:

    def __init__(self, customer_name):
        self.customer_name = customer_name
        self.accounts = []
        self.transactions = []

    def add_account(self, account):
        self.accounts.append(account)

    def make_transaction(self, transaction):
        self.transactions.append(transaction)

    def show_accounts(self):
        for acc in self.accounts:
            print(
                f"Account Holder: {acc._holder_name}, "
                f"Balance: {acc.get_balance()}"
            )


# -----------------------------
# Main Program
# -----------------------------
if __name__ == "__main__":

    # Create Accounts
    savings = SavingsAccount("SB101", "Renov", 10000)
    current = CurrentAccount("CA201", "Renov", 5000)

    # Customer
    customer = Customer("Renov")

    customer.add_account(savings)
    customer.add_account(current)

    # Transactions
    savings.deposit(2000)
    savings.withdraw(3000)

    current.withdraw(8000)

    # Interest
    print("\nSavings Interest:",
          savings.calculate_interest())

    # Store Transaction
    t1 = Transaction("Deposit", 2000)
    customer.make_transaction(t1)

    # Show Details
    print("\nCustomer Accounts:")
    customer.show_accounts()

    print("\nTransaction History:")
    for t in customer.transactions:
        t.show_transaction()

    # Class Method
    print("\nBank Name:",
          Account.get_bank_name())

    # Static Method
    print("Amount Validation:",
          Account.validate_amount(500))
```
